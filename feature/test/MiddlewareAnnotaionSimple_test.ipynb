## 미들웨어

미들웨어를 사용하면 모델 호출 전후에 커스텀 로직을 실행할 수 있습니다. `@before_model` 및 `@after_model` 데코레이터를 사용하여 모델 호출을 감싸는 훅을 정의할 수 있습니다.

**미들웨어 활용 사례:**
- 모델 호출 전 메시지 전처리 (예: 쿼리 재작성)
- 모델 호출 후 응답 후처리 (예: 필터링, 로깅)
- 상태 기반 동적 라우팅

아래 코드는 모델 호출 전후에 로깅을 수행하는 미들웨어 예시입니다.

**미들웨어 훅 요약:**

| 훅 | 시점 | 이 노트북에서 하는 일 |
|:---|:---|:---|
| `before_model` | 모델 호출 직전 | 메시지 개수 로깅, 마지막 사용자 쿼리 재작성 |
| `after_model` | 모델 호출 직후 | 메시지 개수·인덱스별 content 로깅 |

`make_middleware_annotaion_simple_hooks(rewrite_llm=...)` 로 재작성용 LLM을 주입합니다. 메인 응답 모델과 분리하려면 `rewrite_model` 인자를 사용하세요.


```python
from feature.MiddlewareAnnotaionSimple import MiddlewareAnnotaionSimpleAgent
from langchain_core.messages import HumanMessage

workflow = MiddlewareAnnotaionSimpleAgent()
```

```python
# invoke: before_model 재작성 로그 → model → after_model 메시지 출력
_ = workflow.invoke(
    inputs={
        "messages": [
            HumanMessage(content="대한민국 수도")
        ]
    },
)
```

```python
# stream: 동일 미들웨어, 토큰 스트리밍 출력
workflow.stream(
    inputs={
        "messages": [HumanMessage(content="머신러닝 동작 원리 간단히")]
    },
)
```
